# Carregamento dos dados

Nesta etapa são carregados os arquivos CSV do projeto para verificar sua estrutura inicial e confirmar que os dados estão acessíveis.


```python
partidas = pd.read_csv(DATA_PATH / "campeonato-brasileiro-full.csv")
estatisticas = pd.read_csv(DATA_PATH / "campeonato-brasileiro-estatisticas-full.csv")
gols = pd.read_csv(DATA_PATH / "campeonato-brasileiro-gols.csv")
cartoes = pd.read_csv(DATA_PATH / "campeonato-brasileiro-cartoes.csv")
```

```python
datasets = {
  "Partidas": partidas,
  "Estatísticas": estatisticas,
  "Gols": gols,
  "Cartões": cartoes,
}

for nome, df in datasets.items():
  print(f"{nome}: {df.shape[0]} linhas x {df.shape[1]} colunas")
```
